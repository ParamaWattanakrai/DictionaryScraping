Basic Formatting

```python
import pandas as pd

pd.set_option('display.max_colwidth', None)
```

```python
import json

file_path = 'RoyalInstituteDictionary.jsonl'
data = []

with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
            
        entries = json.loads(line)
        
        for entry in entries:
            # Create a flat dictionary for each entry
            row = {
                'word': entry.get('word'),
                'meaning': " ".join(entry.get('means', []))
            }
            
            # Extract 'extra' keys (like 'children', 'parent') into separate columns
            extra = entry.get('extra', {})
            if isinstance(extra, dict):
                for key, value in extra.items():
                    # If the value is a list, join it with commas for the cell
                    if isinstance(value, list):
                        row[key] = ", ".join(value)
                    else:
                        row[key] = value
            
            data.append(row)

df = pd.DataFrame(data)
df
```

```python
import numpy as np

df = df.rename(columns={'': 'reference'})
df = df.replace('', np.nan)
df
```
