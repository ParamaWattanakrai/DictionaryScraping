Merge: G2P + Wikipron

```python
import pandas as pd

g2p = pd.read_csv('g2p-wiktionary-11-2-2020.tsv', sep='\t', names=['writing', 'phonetic'])
g2p = g2p.apply(lambda col: col.str.replace(' ', '', regex=False))
g2p['count'] = g2p['writing'].map(g2p['writing'].value_counts())
g2p
```

```python
wikipron = pd.read_csv('thai_wikipron_5-4-2026.csv')
wikipron
```

```python
g2p[~g2p['writing'].isin(wikipron['writing'])]
```

```python
th_pron = pd.concat([wikipron, g2p])
th_pron = th_pron.drop(columns=['count'])
th_pron = th_pron.drop_duplicates(subset=['writing', 'phonetic']).sort_values(by='writing').reset_index(drop=True)
th_pron['count'] = th_pron['writing'].map(th_pron['writing'].value_counts())
th_pron
```

```python
th_pron.to_csv('th_pron.csv', index=False)
```
