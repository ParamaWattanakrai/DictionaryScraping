```python
import requests
import time
import json

def get_all_thai_words():
    url = "https://en.wiktionary.org/w/api.php"
    params = {
        "action": "query",
        "list": "categorymembers",
        "cmtitle": "Category:Thai lemmas",
        "cmlimit": "max",
        "format": "json",
    }

    headers = {
        "User-Agent": "Polyword Project (contact: Parama.Wattnakrai@example.com)"
    }

    words = []
    batch = 0

    while True:
        r = requests.get(url, params=params, headers=headers)

        try:
            data = r.json()
        except ValueError:
            print("Non-JSON response:", r.status_code, r.text[:200])
            break

        members = data.get("query", {}).get("categorymembers", [])
        words.extend(m["title"] for m in members)

        batch += 1
        print(f"Batch {batch}: +{len(members)} (total: {len(words)})")

        if "continue" in data:
            params.update(data["continue"])
            time.sleep(0.5)
        else:
            print("Done.")
            break

    return words

thai_words = get_all_thai_words()
print(len(thai_words), "Thai words collected")

with open('thai_wiktionary_words.json', 'w', encoding='utf-8') as f:
    json.dump(thai_words, f, ensure_ascii=False, indent=2)
```

```python
import csv
import wikipron

config = wikipron.Config(key='tha')
output_file = 'thai_wiktipron_raw.csv'

with open(output_file, mode='w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)  # default delimiter is ','
    count = 0
    
    for word, pron in wikipron.scrape(config):
        writer.writerow([word, pron])
        f.flush()
        count += 1
        
        if count % 100 == 0:
            print(f'Scraped {count} words...')
    print(f'Total {count} words scraped.')
```
