```python
import pdfplumber

def read_single_page(pdf_path, page_number):
    with pdfplumber.open(pdf_path) as pdf:
        # Indexing is 0-based, so subtract 1 from the desired page number
        page = pdf.pages[page_number - 1]
        text = page.extract_text()
        print(f"--- Content of Page {page_number} ---")
        print(text)

# Execute for page 47
read_single_page("STEDT.pdf", 47)
```
